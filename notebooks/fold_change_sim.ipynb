```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.rcParams.update({'font.size': 14})  # set default font size
```

```python
x = 0.45
y = 0.55

n = 20

r = 3

x1 = [x/n for x in np.random.binomial(n=n, p = x, size = r)]
x2 = [x/n for x in np.random.binomial(n=n, p = y, size = r)]
```

```python
x1
```

```python
df = pd.DataFrame({ "gene expression" : list(x1) + list(x2), "group" : ["group 1"] * r + ["group 2"] * r})
```

```python
color_map = {"group 1" : u'#d62728', "group 2" : u'#9467bd'}
sns.displot(df, x = "gene expression",fill="group", hue = "group",kind="kde",palette=color_map)
plt.axvline(x=np.mean(x1),linestyle="--",color=u'#d62728')
plt.axvline(x=np.mean(x2),linestyle="--",color = u'#9467bd')
plt.xlim(-0.2,1.2)
plt.title("Group 1: {}\nGroup 2: {}\nnumber of reads = {}".format(x, y, n))

plt.show()
```

```python
sns.catplot(df, x = "group", y = "gene expression", hue = "group")
plt.title("Group 1: {}\nGroup 2: {}\nnumber of reads = {}".format(x, y, n))
plt.show()
```
